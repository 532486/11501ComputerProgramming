> Python 物件在 CPython 中是如何實作的？

| 數學概念 | Python 型別 | CPython 實作 | 主要特點 |
|----------|------------|-------------|----------|

請針對第三部分中的每一種型別回答：

1. 它是否為 Python 的內建型別？ 都是
2. 對應的 CPython 物件結構名稱為何？int:PyLongObject、float:PyFloatObject、complex:PyComplexObject、bool:PyBoolObject、list:PyListObject、set:PySetObject
3. 它如何儲存資料？ int:不是固定長度整數，而是由多個數位區塊組成的可變長度結構。float:double。complex:兩個 double。set:Hash Table。bool:True/False
4. 是否有固定大小限制？int:否。float:是。complex:是。set:否。bool:是。list:否
5. 若資料規模持續增加，如何擴充儲存空間？int:配置更多digits。float:無法擴充。complex:無法擴充。set:建立更大的 hash table並重新配置資料。bool:不需要。list:重新配置更大的陣列

## 第五部分：Python 整數與浮點數

執行下列程式：

```python
print(10**100)
print(10**1000)
```

回答：

1. 為什麼 Python 能表示如此大的整數？ 因為 Python 的 int 採用任意精度（Arbitrary Precision）設計，數字變大時可以自動增加儲存空間。
2. Python 的 `int` 是否具有固定的位元數限制？ 沒有
3. 什麼是 Arbitrary Precision（任意精度）？ 指整數的位數不固定，可根據需要動態增加，理論上只受可用記憶體限制。
4. CPython 使用什麼物件來實作 `int`？ PyLongObject
5. Python 整數大小真正的限制是什麼？ 電腦記憶體容量、作業系統可配置記憶體

## 第六部分：AI 模型中的數值表示

人工智慧模型需要儲存大量參數（Parameters）與權重（Weights）。

請查閱資料並回答：

1. AI 模型中的權重通常使用什麼型別儲存？ 通常使用浮點數（Floating Point）儲存
2. 什麼是 FP64？ 雙精度浮點數，它具有最高精度，約可提供 15～17 位十進位有效數字。常用於科學計算，不常用於大型 AI 模型。
3. 什麼是 FP32？ 單精度浮點數，長期以來是深度學習的標準格式，兼顧精度與效能。每個權重大約需要 4 Bytes 儲存空間。。
4. 什麼是 FP16？ 半精度浮點數
5. 什麼是 BF16？ 是 Google 為 AI 訓練提出的格式。
6. 為什麼大型 AI 模型不直接使用 Python 的 `int` 作為權重？ AI 權重需要表示小數且追求高效運算，因此使用浮點數而非 int。

## 第七部分：比較不同數值表示方式

完成下表。

| 表示方式 | 位元數 | 是否有誤差 | 主要用途 |
|----------|--------|------------|----------|

並回答：

2. 哪種表示方式最節省記憶體？ FP16 與 BF16 
3. 哪種表示方式最常見於 AI 訓練？ BF16
4. 哪種表示方式最常見於 AI 推論（Inference）？ FP16

## 第八部分：綜合思考

### 思考題一

Python 官方文件指出：

```python
int
```

具有 Unlimited Precision（無限制精度）。

請回答：

1. Unlimited Precision 是什麼意思？ 整數精度可動態增加，沒有固定位元數限制。
2. 它是否代表可以儲存無限大的整數？ 可以，仍受電腦記憶體限制。
3. CPython 如何實現這項特性？ 使用PyLongObject實作int，透過增加digit和記憶體空間來儲存更大的整數。

---

### 思考題二

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？  因為 Python 的 int 採用任意精度（Arbitrary Precision）設計，在 CPython 中由 PyLongObject 實作。當整數變大時，系統會自動配置更多空間來儲存數字
2. 為什麼 `float` 卻存在範圍限制？ 因為 Python 的 float 在 CPython 中使用 C 語言的 double 實作，而 double 採用 IEEE 754 雙精度浮點數格式，總共只有 64 個位元。
3. 為什麼 `float` 可能產生誤差？ 因為許多十進位小數無法用有限個二進位位元精確表示。

---

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。 int 重視正確性，FP16/BF16 重視效率。
2. 各自的優點與缺點。 int 精確但較耗資源；FP16/BF16 快速省記憶體但有誤差。
3. 為什麼大型語言模型通常選擇較低精度的表示方式？ 因為能節省記憶體並加速訓練。
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？ 會增加記憶體需求、降低速度、提高硬體需求與訓練成本。
